A **Digital Signature System** guarantees the authenticity and integrity of digital messages or documents by leveraging cryptographic techniques. It operates by using a **pair of cryptographic keys**—a **private key** to create a unique signature for a message and a **public key** to validate its authenticity. This process ensures that the message remains **unaltered** during transmission and that it was genuinely sent by the claimed sender.

```python
import rsa
print("RSA module imported successfully!")
```

```python
def generate_keys():
    """Generate RSA key pair (public and private keys)."""
    public_key, private_key = rsa.newkeys(2048)
    return public_key, private_key
```

```python
def sign_message(message, private_key):
    """Sign the message using the private key."""
    message_bytes = message.encode()  # Convert message to bytes
    signature = rsa.sign(message_bytes, private_key, 'SHA-256')
    return signature.hex()  # Convert to hexadecimal for easy display
```

```python
def verify_signature(message, signature_hex, public_key):
    """Verify the digital signature using the public key."""
    try:
        message_bytes = message.encode()
        signature = bytes.fromhex(signature_hex)  # Convert hex back to bytes
        rsa.verify(message_bytes, signature, public_key)
        return True  # Signature is valid
    except rsa.VerificationError:
        return False  # Signature is invalid
    except ValueError:
        return False  # Incorrect signature format
```

```python
if __name__ == "__main__":
    print("Generating RSA keys...")
    public_key, private_key = generate_keys()

    message = "This is a confidential message."
    print("\nOriginal Message:", message)

    # Generate digital signature
    signature = sign_message(message, private_key)
    print("\nGenerated Digital Signature:", signature)

    # Verify signature
    is_valid = verify_signature(message, signature, public_key)
    if is_valid:
        print("\n✅ Signature is valid!")
    else:
        print("\n❌ Signature is invalid!")
```

***RSA Digital Signature Scheme using Python***

The RSA algorithm is a form of asymmetric cryptography, meaning it utilizes a pair of keys: a **public key** and a **private key**. The **public key** is shared openly and can be used by anyone to encrypt messages, while the **private key** remains confidential and is used exclusively for decryption. This ensures secure communication and authentication in digital transactions.

```python
def euclid(m, n):

	if n == 0:
		return m
	else:
		r = m % n
		return euclid(n, r)
```

```python
# Program to find
# Multiplicative inverse
def exteuclid(a, b):

	r1 = a
	r2 = b
	s1 = int(1)
	s2 = int(0)
	t1 = int(0)
	t2 = int(1)

	while r2 > 0:

		q = r1//r2
		r = r1-q * r2
		r1 = r2
		r2 = r
		s = s1-q * s2
		s1 = s2
		s2 = s
		t = t1-q * t2
		t1 = t2
		t2 = t

	if t1 < 0:
		t1 = t1 % a

	return (r1, t1)
```
